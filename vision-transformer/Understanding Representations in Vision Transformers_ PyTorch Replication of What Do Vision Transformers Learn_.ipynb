# 📑 What Do Vision Transformers Learn?

---

## **Abstract**
This paper investigates the internal mechanisms and learned representations of **Vision Transformers (ViTs)**. It explores how ViTs differ from **Convolutional Neural Networks (CNNs)** in feature learning, attention behavior, and interpretability. Using visualization, probing, and representation similarity analysis, the authors reveal both the **strengths** and **limitations** of attention-driven architectures.

---

## **Problem**
While ViTs achieve state-of-the-art results, their inner workings remain poorly understood. The study addresses key questions:

- How do ViTs represent visual features compared to CNNs?  
- What roles do attention heads and layers play in shaping representations?  
- Are ViTs truly capturing global context, or implicitly learning convolution-like locality?  

---

## **Purpose**
The aims are to:

1. Characterize the representations learned by ViTs.  
2. Compare ViTs with CNNs in terms of **hierarchical feature learning** and **inductive biases**.  
3. Enhance interpretability by analyzing attention heads, layers, and token interactions.  

---

## **Methodology**
The authors employ multiple approaches:

- **Visualization of Attention Maps:** Examines token attention patterns across layers → locality vs. global context.  
- **Probing Classifiers:** Linear probes on intermediate features to test semantic richness.  
- **Representation Similarity Analysis (RSA):** Compares ViT vs. CNN representations.  
- **Ablation Studies:** Masking tokens/heads to measure their contribution to accuracy.  

---

## **Results**

- **Locality Emergence:** Early ViT layers learn CNN-like local filters despite lacking convolutional bias.  
- **Global Context Integration:** Middle/deep layers model long-range dependencies better than CNNs.  
- **Layer Specialization:**  
  - Early → textures, edges.  
  - Middle → object parts.  
  - Late → semantic object-level info.  
- **Attention Redundancy:** Many heads are redundant; pruning minimally affects accuracy → overparameterization.  
- **Transferability:** ViT features transfer well across tasks, aligning with semantic rather than spatial similarity.  

---

## **Conclusions**

- ViTs **recover CNN-like locality** but extend it with **global relational reasoning**.  
- Their hierarchical representations mirror CNNs yet remain more flexible in modeling dependencies.  
- Not all attention heads are equally important → potential for **efficient ViT pruning**.  
- The work bridges the **interpretability gap** by clarifying what ViTs learn, showing them as both comparable and complementary to CNNs.  

---

✅ **Essence:**  
ViTs combine **local CNN-style inductive features** with **global attention-driven reasoning**, producing powerful—but often overparameterized—architectures.  


# 📐 Mathematical & Statistical Backbone of *What Do Vision Transformers Learn?*

---

## 1. Feature Activation Maximization
The core loss for visualizing feature **i** at layer **l** over patches **p**:

$$
L_{\text{main}}(x, l, i) = \sum_{p} \big(f_{l,i}\big)_{p}
$$

where $f_{l,i}$ is formed by concatenating the **i-th entry** of patch representations across all patches.

---

## 2. Optimization Objective
Augmented objective with total variation (TV) regularization and input augmentations:

$$
x^{*} = \arg\max_{x} \sum_{k} L_{\text{main}}(a_k(x), l, i) + \lambda \, TV(a_k(x))
$$

- $a_k(x)$: augmented versions of input $x$  
- $\lambda$: regularization hyperparameter  

---

## 3. Augmentation Operators

- **Gaussian Smoothing**
  $$
  GS(x) = x + \varepsilon, \quad \varepsilon \sim \mathcal{N}(0,1)
  $$

- **Color Shift**
  $$
  CS(x) = \sigma x + \mu, \quad \mu \sim U(-1,1), \; \sigma \sim e^{U(-1,1)}
  $$

Here:
- $\varepsilon$ = pixel-level noise  
- $\mu$ = per-channel shift  

---

## 4. Feed-Forward Layer Representation
The **ViT feed-forward block**:

- **Input:** $(b, p, d)$  
- **Projection:** $(b, p, t \cdot d), \; t=4$  
- **Non-linearity:** GELU  
- **Output:** $(b, p, d)$

````python
# ============================================================
# Replication of "What Do Vision Transformers Learn?"
# Compact PyTorch Implementation on CIFAR-10
# ============================================================

!pip install torchinfo
from torchinfo import summary

import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import numpy as np

# -------------------------
# 1. Dataset (CIFAR-10)
# -------------------------
transform = transforms.Compose([
    transforms.Resize(224),   # ViT expects 224x224 input
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5)) # 3 channels
])

trainset = torchvision.datasets.CIFAR10(root='./data', train=True, download=True, transform=transform)
testset  = torchvision.datasets.CIFAR10(root='./data', train=False, download=True, transform=transform)
trainloader = torch.utils.data.DataLoader(trainset, batch_size=64, shuffle=True)
testloader  = torch.utils.data.DataLoader(testset, batch_size=64, shuffle=False)
classes = trainset.classes

# -------------------------
# 2. Vision Transformer
# -------------------------
from torchvision.models import vit_b_16

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
net = vit_b_16(weights=None, num_classes=10).to(device)  # train from scratch

# Model summary
summary(net, input_size=(1, 3, 224, 224))

# -------------------------
# 3. Training
# -------------------------
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(net.parameters(), lr=1e-4)

for epoch in range(2):   # demo epochs
    net.train()
    running_loss = 0.0
    for imgs, labels in trainloader:
        imgs, labels = imgs.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = net(imgs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()
    print(f"Epoch {epoch+1}, Loss: {running_loss/len(trainloader):.4f}")

# -------------------------
# 4. Evaluation
# -------------------------
net.eval()
correct, total = 0, 0
with torch.no_grad():
    for imgs, labels in testloader:
        imgs, labels = imgs.to(device), labels.to(device)
        outputs = net(imgs)
        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()
print(f"Test Accuracy: {100*correct/total:.2f}%")

# -------------------------
# 5. Extract CLS & Patch Embeddings
# -------------------------
embeddings = {}
def hook_fn(module, input, output):
    embeddings['tokens'] = output.detach()

net.encoder.layers[-1].register_forward_hook(hook_fn)

imgs, labels = next(iter(testloader))
imgs, labels = imgs.to(device), labels.to(device)
with torch.no_grad():
    _ = net(imgs)

tokens = embeddings['tokens']   # (batch, num_tokens, dim)
cls_embedding = tokens[:,0,:]      # CLS token
patch_embeddings = tokens[:,1:,:]  # Patch tokens
print("CLS shape:", cls_embedding.shape)
print("Patches shape:", patch_embeddings.shape)

# -------------------------
# 6. Attention Visualization
# -------------------------
attn_weights = {}
def get_attention_map(name):
    def hook(module, input, output):
        attn_weights[name] = output.detach().cpu()
    return hook

# register hook for first two attention blocks
for i, blk in enumerate(net.encoder.layers[:2]):
    blk.attn.attn_drop.register_forward_hook(get_attention_map(f"layer{i}"))

with torch.no_grad():
    _ = net(imgs[:1])  # forward pass for one image

attn_map = list(attn_weights.values())[0][0]  # shape: (heads, tokens, tokens)
plt.imshow(attn_map.mean(0).numpy(), cmap="viridis")
plt.title("Average Attention Map (Layer 0)")
plt.colorbar()
plt.show()

# -------------------------
# 7. Feature Visualization (Neuron Maximization)
# -------------------------
x = torch.randn(1,3,224,224, device=device, requires_grad=True)
optimizer = optim.Adam([x], lr=0.1)

target_neuron = 5  # arbitrary class logit index
for step in range(30):
    optimizer.zero_grad()
    out = net(x)
    loss = -out[0,target_neuron]  # maximize activation
    loss.backward()
    optimizer.step()
    x.data = torch.clamp(x.data, -1, 1)  # keep stable

img = x.detach().cpu().squeeze().permute(1,2,0).numpy()
plt.imshow((img - img.min()) / (img.max() - img.min()))
plt.title(f"Optimized Input for Neuron {target_neuron}")
plt.axis("off")
plt.show()
````

````python
# ============================================================
# Simplified Replication of "What Do Vision Transformers Learn?"
# Fast Demo on CIFAR-10 using ViT-Tiny
# ============================================================

import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
from torchvision.models import vit_b_16

# -------------------------
# 1. Dataset (CIFAR-10)
# -------------------------
transform = transforms.Compose([
    transforms.Resize((224, 224)),   # enforce correct size
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

trainset = torchvision.datasets.CIFAR10(root='./data', train=True, download=True, transform=transform)
testset  = torchvision.datasets.CIFAR10(root='./data', train=False, download=True, transform=transform)
trainloader = torch.utils.data.DataLoader(trainset, batch_size=128, shuffle=True)
testloader  = torch.utils.data.DataLoader(testset, batch_size=128, shuffle=False)

# -------------------------
# 2. Vision Transformer (Tiny)
# -------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
net = vit_b_16(weights="IMAGENET1K_V1")   # load pretrained for speed
net.heads.head = nn.Linear(net.heads.head.in_features, 10)  # replace head
net = net.to(device)

# -------------------------
# 3. Training (Quick Demo)
# -------------------------
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(net.parameters(), lr=1e-4)

for epoch in range(1):   # just 1 epoch demo
    net.train()
    running_loss = 0.0
    for imgs, labels in trainloader:
        imgs, labels = imgs.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = net(imgs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()
    print(f"Epoch {epoch+1}, Loss: {running_loss/len(trainloader):.4f}")

# -------------------------
# 4. Evaluation
# -------------------------
net.eval()
correct, total = 0, 0
with torch.no_grad():
    for imgs, labels in testloader:
        imgs, labels = imgs.to(device), labels.to(device)
        outputs = net(imgs)
        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()
print(f"Test Accuracy: {100*correct/total:.2f}%")
````
